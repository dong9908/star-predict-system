```python
# 값 테스트용.
import pandas as pd

from astropy.coordinates import SkyCoord, EarthLocation, AltAz
from astropy.time import Time
import astropy.units as u


# CSV 불러오기
df = pd.read_csv("data/hyg/current/hyg_v44.csv.gz")


# 오리온자리 별만 추출
stars = df[
    (df["con"] == "Ori") &
    (df["ra"].notna()) &
    (df["dec"].notna())
]


# 첫 번째 별 선택
star = stars.iloc[0]

ra = star["ra"]
dec = star["dec"]


# 별의 적경 / 적위
star_coord = SkyCoord(
    ra=ra * u.deg,
    dec=dec * u.deg,
    frame="icrs"
)


# 관측 위치: 서울
location = EarthLocation(
    lat=37.5665 * u.deg,
    lon=126.9780 * u.deg,
    height=38 * u.m
)


# 현재 시간
time = Time.now()


# 관측 좌표계
altaz_frame = AltAz(
    obstime=time,
    location=location
)


# 적경/적위를 현재 하늘의 고도/방위각으로 변환
altaz = star_coord.transform_to(altaz_frame)


# 결과 출력
print("========== Astropy 테스트 ==========")
print("별자리 :", star["con"])
print("ID     :", star["id"])
print("RA     :", ra, "도")
print("Dec    :", dec, "도")
print("Mag    :", star["mag"])
print()
print("현재 고도(Altitude):", altaz.alt.deg, "도")
print("현재 방위각(Azimuth):", altaz.az.deg, "도")
```
